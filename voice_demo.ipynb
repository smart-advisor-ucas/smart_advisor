# Voice Demo (Real Mode)

Runs `transcribe_audio` (STT) and `synthesize_speech` (TTS) from `src/voice/` against the **real** backends (Whisper + Azure/Piper) — not the mock. Run top to bottom.

```python
import os
import sys
from pathlib import Path

REPO_ROOT = Path.cwd()
SRC_PATH = REPO_ROOT / "src"
if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

os.environ["VOICE_BACKEND"] = "real"  # must be set before importing voice

import voice
from voice import transcribe_audio, synthesize_speech, VoiceError

voice.health_check()
```

## 1. Speech → Text (STT)

```python
AUDIO_PATH =  r'C:\Users\sajaa\OneDrive\Desktop\college\4th year\Second semester\Graduation project 2\smart-advisor-repo\data\test_audio\ما هو الفرق بين تخصص الذكاء الاصطناعي وتخصص الأمن السيبراني.wav'

# If AUDIO_PATH doesn't exist, drop a short Arabic .wav or .m4a at
# data/test_audio/sample_ar.wav (or point AUDIO_PATH at your own file) and re-run this cell.

try:
    stt_result = transcribe_audio(AUDIO_PATH)
    print("Text:", stt_result.text)
    print("Backend:", stt_result.backend)
except VoiceError as e:
    print(f"STT failed ({type(e).__name__}): {e}")
```

## 2. Text → Speech (TTS)

```python
TEXT = "يتطلب التخصص إتمام مئة وستة وعشرين ساعة معتمدة للتخرج."

try:
    tts_result = synthesize_speech(TEXT, voice="male")  # "male" -> ar-SA-HamedNeural (see AZURE_VOICES)
    print("Audio path:", tts_result.audio_path)
    print("Backend:", tts_result.backend)
except VoiceError as e:
    print(f"TTS failed ({type(e).__name__}): {e}")
```
